```python
import pandas as pd
# 1. 지역 리스트 정의 (데이터에 있는 정확한 명칭 사용)
districts = ['성남시 수정구', '성남시 중원구', '성남시 분당구']
district_names = ['sujeong', 'jungwon', 'bundang'] # 변수명용 짧은 이름

# 2. 승용차(final_car) 지역별 분할
car_by_dist = {name: final_car[final_car['O_CTY_NM'] == dist].copy()
                for dist, name in zip(districts, district_names)}

# 3. 버스(final_bus) 지역별 분할
bus_by_dist = {name: final_bus[final_bus['O_CTY_NM'] == dist].copy()
                for dist, name in zip(districts, district_names)}

subway_by_dist = {name: final_subway[final_subway['O_CTY_NM'] == dist].copy()
                for dist, name in zip(districts, district_names)}
# [확인 예시] 분당구의 승용차와 버스 데이터 행 수 확인
print(f"분당구 승용차 데이터 수: {len(car_by_dist['bundang'])}")
print(f"분당구 버스 데이터 수: {len(bus_by_dist['bundang'])}")
# 1. 결과를 담을 리스트 초기화
analysis_results = []
districts = ['sujeong', 'jungwon', 'bundang']
display_names = {'sujeong': '수정구', 'jungwon': '중원구', 'bundang': '분당구'}

# 2. 지역별로 순회하며 탄소 배출량 계산
for dist in districts:
    # 각 지역/수단별 탄소 배출량 합계 (단위: kg -> t 변환하여 가독성 높임)
    car_emission = car_by_dist[dist]['CARBON_EMISSIONS'].sum() / 1000  # 톤(t) 단위
    bus_emission = bus_by_dist[dist]['CARBON_EMISSIONS'].sum() / 1000  # 톤(t) 단위

    # 총 이동 횟수(CNT)도 함께 추출하여 효율성 계산
    car_cnt = car_by_dist[dist]['CNT'].sum()
    bus_cnt = bus_by_dist[dist]['CNT'].sum()

    analysis_results.append({
        '지역': display_names[dist],
        '승용차 배출량(t)': car_emission,
        '버스 배출량(t)': bus_emission,
        '승용차 대비 버스 배출비율(%)': (bus_emission / car_emission) * 100 if car_emission > 0 else 0,
        '승용차 건당 배출(kg)': (car_emission * 1000) / car_cnt if car_cnt > 0 else 0,
        '버스 건당 배출(kg)': (bus_emission * 1000) / bus_cnt if bus_cnt > 0 else 0
    })

# 3. 데이터프레임으로 변환 및 출력
df_analysis = pd.DataFrame(analysis_results)

# 소수점 둘째자리까지 포맷팅
pd.options.display.float_format = '{:,.2f}'.format
print("=== 성남시 구별 승용차 vs 버스 탄소 배출 비교 ===")
print(df_analysis)
```

```python
import pandas as pd

# 1. 결과를 담을 리스트 초기화
recovery_analysis = []
districts = ['sujeong', 'jungwon', 'bundang']
display_names = {'sujeong': '수정구', 'jungwon': '중원구', 'bundang': '분당구'}

for dist in districts:
    # 데이터 추출
    car_data = car_by_dist[dist]
    bus_data = bus_by_dist[dist]
    subway_data = subway_by_dist[dist]

    # 1) 통행량(CNT) 회수 비율 (수단 점유율)
    total_cnt = car_data['CNT'].sum() + bus_data['CNT'].sum() + subway_data['CNT']
    cnt_recovery_rate = (car_data['CNT'].sum() / total_cnt)

    # 2) 탄소 배출 절감 효율 (승용차 대비 버스의 배출 비중)
    # 이 비율이 낮을수록 버스가 탄소 측면에서 효율적으로 '회수'하고 있다는 의미
    car_emission = car_data['CARBON_EMISSIONS'].sum()
    bus_emission = bus_data['CARBON_EMISSIONS'].sum()
    subway_emission = subway_data['CARBON_EMISSIONS'].sum()
    emission_ratio = (car_emission/bus_emission+subway_emission)

    # 3) 거리당 탄소 회수 효율 (Distance-based Efficiency)
    # (버스 이용 거리 합 / 승용차 이용 거리 합)
    car_total_dist = (car_data['CNT'] * car_data['DISTANCE']).sum()
    bus_total_dist = (bus_data['CNT'] * bus_data['DISTANCE']).sum()
    subway_total_dist = (subway_data['CNT'] * subway_data['DISTANCE']).sum()
    dist_recovery_rate = (car_total_dist / bus_total_dist+subway_total_dist)
    recovery_analysis.append({
        '지역': display_names[dist],
        '통행 회수율(점유율, %)': cnt_recovery_rate,
        '거리 회수율(%)': dist_recovery_rate,
        '탄소 배출 비중(%)': emission_ratio,
        '배출 저감 지수': cnt_recovery_rate / emission_ratio
    })

# 2. 데이터프레임 변환
df_recovery = pd.DataFrame(recovery_analysis)

# 3. 출력
pd.options.display.float_format = '{:,.2f}'.format
print("=== 성남시 지역별 대중교통(버스) 회수 효율 분석 ===")
print(df_recovery)
```

```python
import pandas as pd

# 1. 결과를 담을 리스트 초기화
recovery_analysis = []
districts = ['sujeong', 'jungwon', 'bundang']
display_names = {'sujeong': '수정구', 'jungwon': '중원구', 'bundang': '분당구'}

for dist in districts:
    # 데이터 추출
    car_data = car_by_dist[dist]
    bus_data = bus_by_dist[dist]

    # 1) 통행량(CNT) 회수 비율 (수단 점유율)
    total_cnt = car_data['CNT'].sum() + bus_data['CNT'].sum()
    cnt_recovery_rate = (bus_data['CNT'].sum() / total_cnt) * 100 if total_cnt > 0 else 0

    # 2) 탄소 배출 절감 효율 (승용차 대비 버스의 배출 비중)
    # 이 비율이 낮을수록 버스가 탄소 측면에서 효율적으로 '회수'하고 있다는 의미
    car_emission = car_data['CARBON_EMISSIONS'].sum()
    bus_emission = bus_data['CARBON_EMISSIONS'].sum()
    emission_ratio = (bus_emission / car_emission) * 100 if car_emission > 0 else 0

    # 3) 거리당 탄소 회수 효율 (Distance-based Efficiency)
    # (버스 이용 거리 합 / 승용차 이용 거리 합)
    car_total_dist = (car_data['CNT'] * car_data['DISTANCE']).sum()
    bus_total_dist = (bus_data['CNT'] * bus_data['DISTANCE']).sum()
    dist_recovery_rate = (bus_total_dist / car_total_dist) * 100 if car_total_dist > 0 else 0

    recovery_analysis.append({
        '지역': display_names[dist],
        '통행 회수율(점유율, %)': 100-cnt_recovery_rate,
        '거리 회수율(%)': 100-dist_recovery_rate,
        '탄소 배출 비중(%)': 100-emission_ratio,
        '배출 저감 지수': 100-cnt_recovery_rate / emission_ratio if emission_ratio > 0 else 0
    })

# 2. 데이터프레임 변환
df_recovery = pd.DataFrame(recovery_analysis)

# 3. 출력
pd.options.display.float_format = '{:,.2f}'.format
print("=== 성남시 지역별 대중교통(버스) 회수 효율 분석 ===")
print(df_recovery)
```

```python
# 결과를 담을 딕셔너리 구조 생성
district_mapping = {
    'sujeong': '성남시 수정구',
    'jungwon': '성남시 중원구',
    'bundang': '성남시 분당구'
}

# 최종 결과를 저장할 변수들
final_admi_car = {}
final_admi_bus = {}
final_admi_subway = {}

for key, full_name in district_mapping.items():
    # 1. 승용차(Car) 행정동별 배분
    temp_car = cars[cars["O_CTY_NM"] == full_name]
    car_ratio = temp_car['O_ADMI_NM'].value_counts() / temp_car['O_ADMI_NM'].value_counts().sum()
    final_admi_car[key] = car_ratio * car_by_dist[key]['CARBON_EMISSIONS'].sum()

    # 2. 버스(Bus) 행정동별 배분
    temp_bus = buses[buses["O_CTY_NM"] == full_name]
    bus_ratio = temp_bus['O_ADMI_NM'].value_counts() / temp_bus['O_ADMI_NM'].value_counts().sum()
    final_admi_bus[key] = bus_ratio * bus_by_dist[key]['CARBON_EMISSIONS'].sum()

    # 3. 지하철(Subway) 행정동별 배분
    temp_subway = subways[subways["O_CTY_NM"] == full_name]
    subway_ratio = temp_subway['O_ADMI_NM'].value_counts() / temp_subway['O_ADMI_NM'].value_counts().sum()
    # 주의: subway_by_dist 딕셔너리가 미리 정의되어 있어야 합니다.
    final_admi_subway[key] = subway_ratio * subway_by_dist[key]['CARBON_EMISSIONS'].sum()

# [확인] 분당구의 행정동별 버스 탄소 배출량 상위 5개동
print("=== 분당구 행정동별 버스 탄소 배출 추정치 ===")
print(final_admi_bus['bundang'].head())
```

```python
import pandas as pd

# 1. 결과를 담을 리스트 초기화
ratio_results = []

for key, full_name in district_mapping.items():
    # 해당 구의 행정동 리스트 추출
    admi_list = final_admi_car[key].index.union(final_admi_bus[key].index).union(final_admi_subway[key].index)

    for admi in admi_list:
        # 각 수단별 배출량 (데이터가 없는 경우 0 처리)
        car_e = final_admi_car[key].get(admi, 0)
        bus_e = final_admi_bus[key].get(admi, 0)
        subway_e = final_admi_subway[key].get(admi, 0)

        public_total_e = bus_e + subway_e

        # 비율 계산 (대중교통 배출량이 0인 경우를 대비해 처리)
        # 비율 = 승용차 배출량 / (버스 + 지하철 배출량)
        ratio = car_e / (public_total_e+car_e) if public_total_e > 0 else None

        ratio_results.append({
            '시군구명': full_name,
            '행정동명': admi,
            '배출량_비율(차/대중교통)': ratio
        })

# 2. 데이터프레임 생성
df_final_ratio = pd.DataFrame(ratio_results)

# 3. CSV 파일로 저장
file_name = '성남시_행정동별_탄소배출_비율_분석.csv'
df_final_ratio.to_csv(file_name, index=False, encoding='utf-8-sig')

print(f"=== {file_name} 저장 완료 ===")
print(df_final_ratio.head())
```
